```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

DATA_DIR = Path(r"C:\Users\aduak\Downloads\capstone")
CLEAN_DIR = DATA_DIR / "cleaned"

cweeds = pd.read_csv(
    CLEAN_DIR / "clean_cweeds_4cities.csv",
    parse_dates=["timestamp"],
    low_memory=False
)

print("Shape:", cweeds.shape)
print("Date range:", cweeds["timestamp"].min(), "to", cweeds["timestamp"].max())

display(cweeds.head())
```

```python
model_df = cweeds[
    [
        "timestamp",
        "city",
        "latitude",
        "longitude",
        "elevation_m",

        "ghi_wh_m2",

        "dry_bulb_temp_c",
        "dew_point_temp_c",
        "wind_speed_ms",
        "station_pressure_pa"
    ]
].copy()

print("Model dataset shape:", model_df.shape)

display(model_df.head())
```

```python
model_df["year"] = model_df["timestamp"].dt.year
model_df["month"] = model_df["timestamp"].dt.month
model_df["day"] = model_df["timestamp"].dt.day
model_df["day_of_year"] = model_df["timestamp"].dt.dayofyear
model_df["hour"] = model_df["timestamp"].dt.hour

print("Calendar features created.")

display(
    model_df[
        [
            "timestamp",
            "year",
            "month",
            "day",
            "day_of_year",
            "hour"
        ]
    ].head(10)
)
```

```python
# Hour-of-day cyclical encoding

model_df["hour_sin"] = np.sin(
    2 * np.pi * model_df["hour"] / 24
)

model_df["hour_cos"] = np.cos(
    2 * np.pi * model_df["hour"] / 24
)


# Annual seasonal cyclical encoding

model_df["doy_sin"] = np.sin(
    2 * np.pi * model_df["day_of_year"] / 365.25
)

model_df["doy_cos"] = np.cos(
    2 * np.pi * model_df["day_of_year"] / 365.25
)

display(
    model_df[
        [
            "timestamp",
            "hour",
            "hour_sin",
            "hour_cos",
            "day_of_year",
            "doy_sin",
            "doy_cos"
        ]
    ].head(24)
)
```

```python
print("Rows:", f"{len(model_df):,}")
print("Columns:", model_df.shape[1])

print("\nMissing values:")
print(
    model_df.isna()
    .sum()
    .sort_values(ascending=False)
)
```

```python
print("\nColumn types:")
print(model_df.dtypes)
```

```python
model_df = pd.get_dummies(
    model_df,
    columns=["city"],
    prefix="city",
    dtype=int
)

print("City encoding completed.")

display(model_df.head())
```

```python
print(model_df.columns.tolist())
```

```python
feature_columns = [
    # Geography
    "latitude",
    "longitude",
    "elevation_m",

    # Weather
    "dry_bulb_temp_c",
    "dew_point_temp_c",
    "wind_speed_ms",
    "station_pressure_pa",

    # Cyclical time
    "hour_sin",
    "hour_cos",
    "doy_sin",
    "doy_cos",

    # City indicators
    "city_Calgary",
    "city_Edmonton",
    "city_Lethbridge",
    "city_Medicine Hat"
]

target_column = "ghi_wh_m2"

X = model_df[feature_columns].copy()
y = model_df[target_column].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)

display(X.head())
```

```python
train_mask = model_df["year"] <= 2011

validation_mask = (
    (model_df["year"] >= 2012) &
    (model_df["year"] <= 2014)
)

test_mask = model_df["year"] >= 2015


X_train = X.loc[train_mask]
y_train = y.loc[train_mask]

X_val = X.loc[validation_mask]
y_val = y.loc[validation_mask]

X_test = X.loc[test_mask]
y_test = y.loc[test_mask]
```

```python
split_summary = pd.DataFrame({
    "Split": ["Train", "Validation", "Test"],
    "Rows": [
        len(X_train),
        len(X_val),
        len(X_test)
    ],
    "Start_Year": [
        model_df.loc[train_mask, "year"].min(),
        model_df.loc[validation_mask, "year"].min(),
        model_df.loc[test_mask, "year"].min()
    ],
    "End_Year": [
        model_df.loc[train_mask, "year"].max(),
        model_df.loc[validation_mask, "year"].max(),
        model_df.loc[test_mask, "year"].max()
    ]
})

split_summary["Percent"] = (
    split_summary["Rows"] /
    len(model_df) * 100
).round(2)

display(split_summary)
```

```python
split_city_check = pd.DataFrame({
    "Train": model_df.loc[train_mask].filter(like="city_").sum(),
    "Validation": model_df.loc[validation_mask].filter(like="city_").sum(),
    "Test": model_df.loc[test_mask].filter(like="city_").sum()
})

display(split_city_check)
```
